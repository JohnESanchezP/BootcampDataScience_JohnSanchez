### Activamos la libreria ###

```python
import requests
import pandas as pd
```

### 1. Extracion de datos de API ###

```python

# 1. Lista para almacenar la información de cada Pokémon
lista_pokemon = []

print("Iniciando la extracción de datos desde la API...")

# 2. Consultamos los primeros 100 Pokémon
for i in range(1, 101):
    url = f"https://pokeapi.co/api/v2/pokemon/{i}"
    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()
        
        # Extraemos características específicas del JSON
        info_pokemon = {
            "id": data["id"],
            "name": data["name"],
            "height": data["height"],  # Altura
            "weight": data["weight"],  # Peso
            "base_experience": data["base_experience"],  # Experiencia base
            "type": data["types"][0]["type"]["name"],  # Tipo principal
            "hp": data["stats"][0]["base_stat"],
            "attack": data["stats"][1]["base_stat"],
            "defense": data["stats"][2]["base_stat"],
            "speed": data["stats"][5]["base_stat"]
        }
        lista_pokemon.append(info_pokemon)
    else:
        print(f"Error al consultar el ID {i}")

print(f"¡Extracción completada con éxito! Se obtuvieron {len(lista_pokemon)} registros.")
```

### 2. Conversión a DataFrame, Limpieza y Transformación ###


```python
# ahora estructuramos los datos con Pandas y creamos variables calculadas para el análisis.

# 1. Convertimos la lista de diccionarios a un DataFrame
df_pokemon = pd.DataFrame(lista_pokemon)

# 2. Transformamos los nombres para que la primera letra sea mayúscula
df_pokemon['name'] = df_pokemon['name'].str.capitalize()
df_pokemon['type'] = df_pokemon['type'].str.capitalize()

# 3. Creamos una columna con el "Total de Stats de Combate"
df_pokemon['total_stats'] = df_pokemon['hp'] + df_pokemon['attack'] + df_pokemon['defense'] + df_pokemon['speed']

# 4. Aseguramos que no haya nulos y revisamos el tipo de datos
df_pokemon = df_pokemon.dropna()

# Mostramos las primeras 5 filas para verificar el resultado
df_pokemon.head()
```

### Exportar a CSV ###

```python
df_pokemon.to_csv("pokemon_data.csv", index=False)
print("¡Archivo 'pokemon_data.csv' guardado y listo para Power BI!")
```

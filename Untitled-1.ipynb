```python
import pandas as pd
import numpy as np
```

**VENTAS TOTAL POR MES**

```python

datos_excel["Valor subtotal "].sum()
```

```python
datos_excel["Fecha"] = pd.to_datetime(datos_excel["Fecha"])
datos_excel["Mes"] = datos_excel["Fecha"]. dt.to_period ("M")
ventas_mensuales = datos_excel.groupby(datos_excel["Fecha"].dt.to_period("M"))["Valor subtotal "].sum()
ventas_mensuales
```

**¿CUAL ES EL PROMEDIO MENSUAL?**

```python
ventas_mensuales.mean()
```

**¿PUEDES MOSTRARME LAS VENTAS ANUALES DE MANERA DESENDIENTES ?**


```python
meses_mas_rentables = datos_excel.groupby(datos_excel["Fecha"].dt.to_period("M"))["Valor subtotal "].sum()
meses_mas_rentables = datos_excel.groupby(datos_excel["Fecha"].dt.to_period("M"))["Valor subtotal "].sum().sort_values(ascending=False)
meses_mas_rentables
```

**MUESTRAME LOS 5 PRODUCTOS MAS VENDIDOS Y QUE HAN GENERADO MAYORES INGRESOS EN TODO EL AÑO?**

```python

datos_excel.groupby(['Desc. item'])[['Cantidad inv.','Valor subtotal ']].sum()

```

```python
top_5 = datos_excel.groupby(['Desc. item'])[['Cantidad inv.','Valor subtotal ']].sum()
top_5 = datos_excel.groupby(['Desc. item'])[['Cantidad inv.', 'Valor subtotal ']].sum().nlargest(6, 'Valor subtotal ')

print(top_5)
```

```python

top_5 = datos_excel.groupby(['Desc. item'])[['Cantidad inv.', 'Valor subtotal ']].sum().nlargest(6, 'Valor subtotal ')
top_5 = top_5.drop('PROPINAS')

print(top_5)
```

**CALCULA EL TICKE PROMEDIO** 

```python
total_ventas = datos_excel['Valor subtotal '].sum()
cantidad_facturas = len(datos_excel)
ticket_promedio = total_ventas / cantidad_facturas

print(ticket_promedio)
```
